#limpieza de datos 

```python
df = df.drop(columns=['Animal code'])
```

```python
df = df.drop(columns=['Animal name'])
```

```python
df = df.drop_duplicates()
```

```python
# se verifica que en la columna Animal type hay valores nulos, por lo que se pricede a cambiar por Unknown.

df['Animal type'] = df['Animal type'].fillna('Unknown')
```

```python
#se verifica cuantos valores nulos hay  y que porcentaje representa del total de los datos.

resumen_nulos = pd.DataFrame({
    'Total Nulos': df.isna().sum(),
    'Porcentaje (%)': (df.isna().mean() * 100).round(2)
})
print(resumen_nulos)
```

```python
#en base al resultado anterior se valida que la columna con mayor cantidad de valores nulos es Animal name
#se eliminan las filas que tengan 5 o mas valores nulos, ya que representan un porcentaje alto de informacion faltante.
df = df.dropna(thresh=5)
```

```python
#ahora se procede a llenar los valores nulos de las columnas Animal name, Gender y Animal type con valores por defecto.
df['Gender'] = df['Gender'].fillna('not determined')
df['Animal type'] = df['Animal type'].fillna('Unknown')
```

```python
df.head(10)
```

```python
#ahora se procede a limpiar los valores paracitos que estan en Animal type

df['Animal type'] = df['Animal type'].astype(str).str.replace('*', '', regex=False).str.strip()
```

```python
#ahora se procede a reindenxar el dataframe para que los indices sean consecutivos y no hayan saltos.
df = df.reset_index(drop=True)
```

```python
df['Animal type'].unique()
```

```python
#se verifican los datos que no estan estandarizados 

print("Valores en Animal type:")
print(df['Animal type'].unique())

print("\nValores en Gender:")
print(df['Gender'].unique())

print("\nValores en Country:")
print(df['Country'].unique())
```

```python
#estandarizacion de datos 

# 1. Quitar espacios invisibles en blanco al inicio o final en todas las columnas de texto
columns_text = ['Animal type', 'Country', 'Gender']
for col in columns_text:
    df[col] = df[col].astype(str).str.strip()

# 2. Diccionario de estandarización para 'Animal type'
map_animals = {
    'European bison*': 'European bison',
    'European bison™': 'European bison',
    'European biscon': 'European bison',
    'European bisson': 'European bison',
    'European buster': 'European bison',  # Por el peso de 600+ kg se confirma que es bisonte
    'lynx?': 'Lynx',
    'red squirrell': 'Red Squirrel',
    'red squirel': 'Red Squirrel',
    'wedgehod': 'Hedgehog',
    'lodgehod': 'Hedgehog'
}
df['Animal type'] = df['Animal type'].replace(map_animals)

# 3. Diccionario de estandarización para 'Country' 
map_country = {
    'PL': 'Poland',
    'DE': 'Germany',
    'HU': 'Hungary',
    'Hungry': 'Hungary',
    'CZ': 'Czech Republic',
    'CC': 'Czech Republic',
    'Czech': 'Czech Republic',
    'Australia': 'Austria',  # Es fauna europea; Australia es una errata por Austria
}
df['Country'] = df['Country'].replace(map_country)
```

```python

#verificacion de estandarizacion de datos

print("Animal type limpio:")
print(df['Animal type'].unique())

print("\nCountry limpio:")
print(df['Country'].unique())

print("\nGender:")
print(df['Gender'].unique())
```

```python
# convertir la fecha a formato datetime

df['Observation date']= pd.to_datetime(df['Observation date'], errors='coerce', format='%Y-%m-%d', dayfirst=True)
```

#NumPy

```python
# se crea una nueva columna llamada Weight Category que clasifica a los animales en dos categorias: Large y Small/Medium, dependiendo de su peso en kg. Si el peso es mayor o igual a 100 kg, se clasifica como Large, de lo contrario se clasifica como Small/Medium.

df['Weight Category'] = np.where( df['Weight kg']>= 100,'Large', 'Small/Medium')

#verificacion de la nueva columna Weight Category
df[['Animal type', 'Weight kg', 'Weight Category']].head(10)
```

```python
mediana_peso = np.nanmedian(df['Weight kg'])
print(f"Mediana de peso: {mediana_peso:.2f} kg")

# representa el percentil 50: el 50 % de los animales observados tiene un peso igual o menor a este valor, siendo una medida más robusta frente a especies con pesos desproporcionados.
```

GROUPBY()

```python
# Conteo de registros por especie (observaciones por animal)
observaciones_por_animal = df.groupby('Animal type').size()
print(observaciones_por_animal)

```

```python
# Promedio de peso por especie (variable numerica segun el tipo de animal)
promedio_peso_animal = df.groupby('Animal type')['Weight kg'].mean()
print(promedio_peso_animal.round(2))
```

```python
# Promedio de longitud corporal por país (variable numerica segun el pais)
promedio_longitud_pais = df.groupby('Country')['Body Length cm'].mean()
print(promedio_longitud_pais.round(2))
```

```python
# Promedio de peso por especie y género
peso_especie_genero = df.groupby(['Animal type', 'Gender'])['Weight kg'].mean()
print(peso_especie_genero.round(2))
```

Evidencia: Al contrastar np.nanmean() frente a np.nanmedian(), la media aritmética resulta significativamente mayor que la mediana. Asimismo, la variable condicional creada con np.where() (Size_Category) evidencia que el peso está polarizado entre megafauna (bisontes que superan los 500 kg) y fauna menor (erizos y ardillas con pesos inferiores a 1 kg).

Qué significa: En este conjunto de datos, el promedio general no es una métrica representativa del "animal típico". Para análisis exploratorios o inferenciales futuros, la mediana es la medida de tendencia central adecuada al ser robusta frente a los valores atípicos que introduce la megafauna.

Conclusión 2: Evidencia de dimorfismo sexual morfológico. Existe una diferencia biométrica consistente en el peso de los animales asociada directamente a su género biológico.

Evidencia: En el análisis multivariable (df.groupby(['Animal type', 'Gender'])['Weight kg'].mean()), los machos (male) registraron un peso promedio notablemente superior al de las hembras (female), observándose la brecha más amplia en el European bison (donde la diferencia alcanza decenas de kilogramos).

Qué significa: Las especies monitoreadas presentan dimorfismo sexual evidente. En términos ecológicos y analíticos, no se debe modelar o evaluar el estado de salud y nutrición de estas poblaciones usando rangos de peso globales, sino que resulta mandatorio segmentar las métricas por sexo.

Conclusión 3: No viabilidad analítica de la identificación individual (Animal name). La variable cualitativa Animal name carece de completitud suficiente para fungir como identificador o factor explicativo en el estudio.

Evidencia: La verificación de calidad inicial reveló que más del 95 % de los registros presentaban valores nulos (NaN), superando ampliamente el umbral tolerable para cualquier análisis estadístico estándar.

Qué significa: En proyectos de monitoreo de fauna silvestre, el levantamiento se enfoca en el avistamiento biológico de poblaciones generales y no en el seguimiento de ejemplares bautizados o domesticados. Esta columna debe descartarse para análisis numéricos o de agrupación, ya que imputarla con 'Unknown' solo tiene fines estéticos y no añade valor  analítico.

Conclusión 4: Deficiencia en los controles de captura de datos taxonómicos y geográficos, La base de datos original presentó una alta tasa de "ruido" e inconsistencias por captura manual no controlada en variables categóricas críticas (Animal type y Country).

Evidencia: La exploración con .unique() arrojó redundancias causadas por erratas tipográficas ('European buster', 'wedgehod'), caracteres parasitos ('European bison*') y una mezcla desordenada entre códigos ISO ('PL', 'DE') y nombres completos ('Poland', 'Germany'), además de inconsistencias geográficas severas como registrar 'Australia' en un ecosistema de fauna europea.

Qué significa: La recolección de datos primaria carecía de validaciones a la entrada (como listas desplegables o reglas de validación en campo). Sin el proceso de limpieza y mapeo con diccionarios de reemplazo realizado en Pandas, cualquier agrupación por especie o país habría fragmentado la muestra en grupos ficticios, invalidando los cálculos de medias y conteos.
