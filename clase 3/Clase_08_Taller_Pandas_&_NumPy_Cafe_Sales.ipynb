## 2. Detectar valores inválidos

Un dato problemático no siempre aparece como `NaN`.

En esta base también podemos encontrar valores como:

```text

```

### Pregunta 5

Busca en qué columnas aparecen estos valores y cuántos hay.

````python
# Tu código aquí
# Pista: puedes recorrer las columnas con un for
#NaN es un valor especial que representa la ausencia de un valor. En Pandas, NaN se utiliza para representar valores faltantes o nulos en un DataFrame.

for column in df.columns:
    error = df[column].isin(["ERROR","UNKNOWN" ]).sum()
    if error > 0:
        print(f"La columna {column} tiene {error} valores de error.")
````

### Pregunta 6

````python
# Tu código aquí
#este proceso reemplaza los valores "ERROR" y "UNKNOWN" por NaN en todo el DataFrame, lo que permite un análisis más preciso de los datos y facilita la limpieza del conjunto de datos.

df = df.replace(["ERROR", "UNKNOWN"], np.nan)
````

- `Quantity`
- `Price Per Unit`

y utiliza `errors="coerce"`.

## 6. NumPy: mediana de `Quantity`

### Pregunta 13

Calcula la mediana de `Quantity` **ignorando los valores nulos**.

Busca una función de NumPy que empiece por:

```python
np.nan...
```

````python
# Tu código aquí

mediana_quantity = np.nanmedian(df['Quantity'])
#np.nanmedian() es una función de NumPy que calcula la mediana de un array, ignorando los valores NaN. En este caso, se está calculando la mediana de la columna 'Quantity' del DataFrame df, que puede contener valores NaN.
mediana_quantity
````

### Pregunta 14

Utiliza la mediana calculada para completar los valores faltantes de `Quantity`.

Después comprueba que ya no existan valores nulos en esa columna.

````python
# Tu código aquí
df['Quantity'] = df['Quantity'].fillna(mediana_quantity)
df['Quantity'].isnull().sum()   
````

Calcula la mediana de `Price Per Unit` utilizando NumPy y úsala para completar cualquier valor faltante.

## 9. Variables categóricas

`Payment Method` y `Location` son variables categóricas.

### Pregunta 17

Si todavía existen valores nulos en estas columnas:

1. Calcula la moda.
2. Utiliza la moda para completar los valores faltantes.
3. Comprueba el resultado.

Puedes utilizar un `for` para resolver ambas columnas.

````python
# Tu código 

for col in ['Payment Method', 'Location']:
    
    if df[col].isna().any(): 
        moda = df[col].mode()[0]
        df[col] = df[col].fillna(moda)
print(df[['Payment Method', 'Location']].isnull().sum())


````

## 10. `Transaction Date`

Tenemos valores faltantes en la fecha de transacción.


### Pregunta 18

Calcula una fecha representativa utilizando la fecha central de los registros disponibles y úsala para completar los valores faltantes.

Pista:

1. Elimina temporalmente los `NaT`.
2. Ordena las fechas.
3. Obtén el valor central.
4. Usa `fillna()`.

````python
# Tu código aquí
mediana_date = pd.to_datetime(np.nanmedian(df['Transaction Date'].astype(np.int64)))

#este codigo es para calcular la mediana de la columna 'Transaction Date' del DataFrame df, que puede contener valores NaN. Primero, se convierte la columna a un tipo de datos entero de 64 bits utilizando astype(np.int64),
# luego se calcula la mediana ignorando los valores NaN con np.nanmedian(). Finalmente, se convierte la mediana calculada de nuevo a un tipo de datos datetime utilizando pd.to_datetime
````

````python
df['Transaction Date'] = df['Transaction Date'].fillna(mediana_date)
df['Transaction Date'].isnull().sum()
#este codigo es para reemplazar los valores nulos en la columna 'Transaction Date' con la mediana calculada anteriormente, y luego verifica si todavía hay valores nulos en esa columna.
````

## Reto final

Ahora que la base está limpia, responde estas preguntas utilizando código.

### 1. ¿Cuál es el producto con mayor cantidad vendida?

### 2. ¿Qué producto genera mayor revenue?


**Pista:** Utiliza la herramienta: `groupby()`

````python
# Tu código aquí

# El método groupby() sigue el principio Split-Apply-Combine (Dividir, Aplicar y Combinar): divide los datos en grupos según una o más columnas, aplica una función a cada grupo (como sumar, promediar o contar) y combina los resultados en una nueva estructura.

highest_quantity = df.groupby('Item')['Quantity'].sum().idxmax()

# la funcion idxmax() funciona para buscar el numero mal alto y me devuelva el nombre del item

print ('el producto con mayor cantidad vendida es ', highest_quantity)
````

````python
best_seller = df.groupby('Item')['Revenue_Calculated'].sum().idxmax()

print ('el producto mas vendido es ', best_seller)
````
